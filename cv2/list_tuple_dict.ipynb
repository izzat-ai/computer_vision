- Berilgan ro'yxatdan faqat mushbat sonlarni (0 va undan katta) ajratib olib, yangi ro'yxat shakllantiring

```python
raw_data = [120, -5, None, 300, 0, -50, None, 450, 210]
new_lst = []
for i in raw_data:
  if i is not None and i >= 0:
    new_lst.append(i)

print("Yangi ro'yxat shakillantirildi:\n", new_lst)
```

1. Ro'yxatdan faqat 0 dan katta bo'lgan to'g'ri narxlarni ajratib oling
2. Qolgan to'g'ri narxlarni eng arzonidan eng qimmatiga qarab (o'sish tartibida) tartiblang va ekranga chiqaring

```python
prices = [45000, -1200, 15000, 89000, 0, 32000, 15000, 5000]
narxlar = []
for price in prices:
  if price > 0:
    narxlar.append(price)

print("to'g'ri narxlar:", sorted(narxlar))
```

1. Ushbu ro'yxatni aylanib (for sikli yordamida), faqat 80 balldan yuqori to'plagan foydalanuvchilarning ismlarini yangi listga yig'ing
2. Natijani ekranga chiqaring

```python
users = [("Ali", 75), ("Otabek", 92), ("Madina", 85), ("Zarina", 60), ("Jasur", 98)]
foydalanuvchilar = []
for user in users:
  if user[1] > 80:
    foydalanuvchilar.append(user[0])

foydalanuvchilar
```

1. Berilgan ro'yxatni aylanib, faqat "car" (mashina) deb topilgan obyektlarni ajratib oling
2. Ushbu mashinalarning eni (width, ya'ni tuple'dagi 3-element) qiymatlarini yangi list ko'rinishida yig'ing va ularning umumiy yig'indisini hisoblang

```python
detections = [("car", (50, 100, 220, 120)), ("person", (10, 20, 40, 90)), ("car", (300, 250, 180, 100)), ("bike", (80, 90, 60, 60)), ("car", (400, 150, 250, 130))]
new_detects = []
for i in detections:
  if i[0] == 'car':
    new_detects.append(i[1][2])

print("Car obyektlarini eni:", new_detects)
print("Ularni yig'indisi:", sum(new_detects))
```

- Agar confidence (aniqlik darajasi) 0.8 dan katta bo'lsa, ekranga "Tasdiqlangan obyekt: [obyekt_nomi]" deb chiqaring. Aks holda "Ishonchsiz obyekt" deb chiqaring

```python
detection = {"label": "person", "confidence": 0.88, "box": (10, 20, 50, 100)}
if detection['confidence'] > 0.8:
  print(f"Tasdiqlangan obyekt: {detection['label']}")
else:
  print("Ishonchsiz obyekt")
```
