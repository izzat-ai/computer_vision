- Berilgan ro'yxatdan faqat mushbat sonlarni (0 va undan katta) ajratib olib, yangi ro'yxat shakllantiring

```python
raw_data = [120, -5, None, 300, 0, -50, None, 450, 210]
new_lst = []
for i in raw_data:
  if i is not None and i >= 0:
    new_lst.append(i)

print("Yangi ro'yxat shakillantirildi:\n", new_lst)
```

- Ushbu ro'yxatdagi eng katta va eng kichik qiymatni toping hamda ular o'rtasidagi farqni (ayirmasini) hisoblang

```python
scores = [88, 12, 95, 43, 67, 100, 25]
print("max qiymat:", max(scores))
print("min qiymat:", min(scores))
print("max va min qiymatlar ayirmasi:", max(scores)-min(scores))
```
