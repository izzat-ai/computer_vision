- Berilgan ro'yxatdan faqat mushbat sonlarni (0 va undan katta) ajratib olib, yangi ro'yxat shakllantiring

```python
raw_data = [120, -5, None, 300, 0, -50, None, 450, 210]
new_lst = []
for i in raw_data:
  if i is not None and i >= 0:
    new_lst.append(i)

print("Yangi ro'yxat shakillantirildi:\n", new_lst)
```

- Ushbu ro'yxatdagi eng katta va eng kichik qiymatni toping hamda ular o'rtasidagi farqni (ayirmasini) hisoblang

```python
scores = [88, 12, 95, 43, 67, 100, 25]
print("max qiymat:", max(scores))
print("min qiymat:", min(scores))
print("max va min qiymatlar ayirmasi:", max(scores)-min(scores))
```

- Ro'yxatdagi takrorlangan elementlarni olib tashlab, faqat bir martadan qoldiring va hosil bo'lgan ro'yxatni o'sish tartibida (kichikdan kattaga) tartiblang

```python
items = [640, 320, 640, 1280, 320, 224, 512, 224]
new_items = set(items)
new_items
```

- Berilgan tuple ichidagi koordinatalardan foydalangan holda obyektning eni (kengligi) va bo'yini (balandligini) hisoblang

```python
box = (50, 100, 250, 400)
eni = box[2] - box[0]
boyi = box[-1] - box[1]
print("eni:", eni)
print("bo'yi:", boyi)
```

- Ushbu tuple ichidagi qiymatlarni bitta qatorda alohida o'zgaruvchilarga ajratib oling va natijani ekranga chiqaring

```python
config = (1280, 720, 30)
olcham1, olcham2, kadr = config
print("olcham1:", olcham1)
print("olcham2:", olcham2)
print("kadr:", kadr)
```
